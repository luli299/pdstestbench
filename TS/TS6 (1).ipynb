circuito a .png

Funcion de transferencia:

$$ Z_1 = j\omega L + R $$

$$ j\omega L + R + \frac{1}{j\omega C} = \frac{-\omega^2 LC + j\omega CR + 1}{j\omega C} $$

$$ V_2 = V_1 \cdot \frac{\dfrac{1}{j\omega C}}{j\omega L + R + \dfrac{1}{j\omega C}} = \frac{V_1}{-\omega^2 LC + j\omega CR + 1} $$

$$ H(\omega) = \frac{V_2}{V_1} = \frac{1}{-\omega^2 LC + j\omega CR + 1} $$

$$ H(s) = \frac{\dfrac{1}{LC}}{s^2 + \dfrac{R}{L}\,s + \dfrac{1}{LC}} $$

Calculo de L y C:

$$ \omega_0^2 = \frac{1}{LC}, \qquad \frac{\omega_0}{Q} = \frac{R}{L}, \qquad Q = \frac{\omega_0 L}{R} = \frac{1}{R}\sqrt{\frac{L}{C}} $$

$$ \sqrt{\frac{L}{C}} = \frac{\sqrt{2}}{2} \;\Rightarrow\; L = \frac{C}{2} $$

$$ \frac{1}{LC} = \frac{2}{C^2} = 1 \;\Rightarrow\; C = \sqrt{2}\ \text{F}, \qquad L = \frac{\sqrt{2}}{2}\ \text{H} $$

$$ \boxed{H(s) = \frac{1}{s^2 + \sqrt{2}\,s + 1}} $$

Polos y ceros:

$$ p_{1,2} = \frac{-\sqrt{2} \pm \sqrt{2 - 4}}{2} = -\frac{1}{\sqrt{2}} \pm j\,\frac{1}{\sqrt{2}} $$

$$ \text{Cero doble en } s \to \infty $$

Modulo y fase:
    
$$ H(\omega) = \frac{1}{(1-\omega^2) + j\sqrt{2}\,\omega} $$

$$ |H(\omega)| = \frac{1}{\sqrt{(1-\omega^2)^2 + 2\omega^2}} = \frac{1}{\sqrt{\omega^4 - 2\omega^2 + 2\omega^2 + 1}} = \frac{1}{\sqrt{\omega^4 + 1}} $$

$$
\angle H(\omega) = -\operatorname{atan2}\!\left(\sqrt{2}\,\omega,\; 1-\omega^2\right) =
\begin{cases}
-\arctan\!\left(\dfrac{\sqrt{2}\,\omega}{1-\omega^2}\right) & \omega < 1 \\[2mm]
-\dfrac{\pi}{2} & \omega = 1 \\[1mm]
-\pi + \arctan\!\left(\dfrac{\sqrt{2}\,\omega}{\omega^2-1}\right) & \omega > 1
\end{cases}
$$

Tipo de filtro y ancho de banda:
$$ |H(\omega)|^2 = \frac{1}{\omega^4 + 1} = \frac{1}{2} \;\Rightarrow\; \omega_c = 1\ \text{rad/s} $$

$$ \boxed{BW = 1\ \text{rad/s}} \qquad (\text{filtro pasabajos}) $$

circuito b.png

Funcion de transferencia:

$$ \frac{1}{Z_1} = j\omega C + \frac{1}{R} = \frac{j\omega CR + 1}{R} \;\Rightarrow\; Z_1 = \frac{R}{j\omega CR + 1}, \qquad Z_2 = j\omega L $$

$$ V_2 = V_1 \cdot \frac{Z_2}{Z_1 + Z_2}, \qquad Z_1 + Z_2 = \frac{R}{j\omega CR + 1} + j\omega L = \frac{R - \omega^2 CRL + j\omega L}{j\omega CR + 1} $$

$$ H(\omega) = \frac{V_2}{V_1} = \frac{j\omega L\,(j\omega CR + 1)}{R - \omega^2 CRL + j\omega L} = \frac{-\omega^2 + \dfrac{1}{CR}\,j\omega}{-\omega^2 + \dfrac{1}{CR}\,j\omega + \dfrac{1}{LC}} $$

$$ H(s) = \frac{s^2 + \dfrac{1}{CR}\,s}{s^2 + \dfrac{1}{CR}\,s + \dfrac{1}{LC}} $$

Calculo de L y C:

$$ \omega_0^2 = \frac{1}{LC} = 1 \;\Rightarrow\; \sqrt{LC} = 1 \;\Rightarrow\; L = \frac{1}{C} $$

$$ \frac{1}{CR} = \frac{\omega_0}{Q} \;\Rightarrow\; CR = \frac{\sqrt{2}}{2} \;\Rightarrow\; C = \frac{\sqrt{2}}{2}\ \text{F} = \frac{1}{\sqrt{2}}\ \text{F}, \qquad L = \frac{2}{\sqrt{2}}\ \text{H} = \sqrt{2}\ \text{H} $$

$$ \boxed{H(s) = \frac{s^2 + \sqrt{2}\,s}{s^2 + \sqrt{2}\,s + 1}} $$

Polos y ceros:

$$ \text{Polos: } p_{1,2} = -\frac{1}{\sqrt{2}} \pm j\,\frac{1}{\sqrt{2}} \qquad \text{Ceros: } z_1 = 0, \quad z_2 = -\sqrt{2} $$

Modulo y fase:

$$ H(\omega) = \frac{-\omega^2 + j\sqrt{2}\,\omega}{(1-\omega^2) + j\sqrt{2}\,\omega} $$

$$ |H(\omega)| = \frac{\sqrt{\omega^4 + 2\omega^2}}{\sqrt{(1-\omega^2)^2 + 2\omega^2}} = \frac{\omega\sqrt{\omega^2 + 2}}{\sqrt{\omega^4 + 1}} $$

$$ |H(1)| = \sqrt{\frac{3}{2}} \approx 1{,}225 $$

$$ \angle H(\omega) = \frac{\pi}{2} + \arctan\!\left(\frac{\omega}{\sqrt{2}}\right) - \operatorname{atan2}\!\left(\sqrt{2}\,\omega,\; 1-\omega^2\right) $$

$$ \lim_{\omega \to \infty} |H(\omega)| = \lim_{\omega \to \infty} \frac{\omega^2\sqrt{1 + \dfrac{2}{\omega^2}}}{\omega^2\sqrt{1 + \dfrac{1}{\omega^4}}} = 1 $$

Tipo de filtro y ancho de banda:

$$ \frac{\omega^4 + 2\omega^2}{\omega^4 + 1} = \frac{1}{2} \;\Rightarrow\; \omega^4 + 4\omega^2 - 1 = 0 \;\Rightarrow\; \omega_c = \sqrt{\sqrt{5} - 2} \approx 0{,}486\ \text{rad/s} $$

$$ \text{Filtro pasa altos: banda de paso } [\omega_c, \infty) $$

Ejercicio c)


circuito c.png

Funcion de transferencia:
    
$$ \frac{1}{Z_1} = \frac{1}{j\omega L} + j\omega C = \frac{1 - \omega^2 CL}{j\omega L} \;\Rightarrow\; Z_1 = \frac{j\omega L}{1 - \omega^2 CL}, \qquad Z_2 = R $$

$$ Z_1 + Z_2 = \frac{j\omega L + R - R\omega^2 CL}{1 - \omega^2 CL} $$

$$ H(\omega) = \frac{Z_1}{Z_1 + Z_2} = \frac{j\omega L}{-RCL\,\omega^2 + j\omega L + R} = \frac{\dfrac{j\omega}{RC}}{-\omega^2 + \dfrac{1}{CR}\,j\omega + \dfrac{1}{CL}} $$

$$ H(s) = \frac{\dfrac{1}{RC}\,s}{s^2 + \dfrac{1}{RC}\,s + \dfrac{1}{LC}} $$

Calculo de L y C:

$$ \omega_0^2 = \frac{1}{CL} = 1 \;\Rightarrow\; L = \frac{1}{C}, \qquad \frac{1}{CR} = \frac{\omega_0}{Q} = \sqrt{2} \;\Rightarrow\; C = \frac{\sqrt{2}}{2}\ \text{F}, \qquad L = \sqrt{2}\ \text{H} $$

$$ \boxed{H(s) = \frac{\sqrt{2}\,s}{s^2 + \sqrt{2}\,s + 1}} $$

Polos y ceros:

$$ \text{Polos: } p_{1,2} = -\frac{1}{\sqrt{2}} \pm j\,\frac{1}{\sqrt{2}} \qquad \text{Ceros: } z_1 = 0, \quad z_2 \to \infty $$

Modulo y fase:

$$ H(\omega) = \frac{j\sqrt{2}\,\omega}{-\omega^2 + j\sqrt{2}\,\omega + 1} $$

$$ |H(\omega)| = \frac{\sqrt{2}\,\omega}{\sqrt{(-\omega^2 + 1)^2 + 2\omega^2}} = \frac{\sqrt{2}\,\omega}{\sqrt{\omega^4 + 1}}, \qquad |H(1)| = 1 $$

$$ \angle H(\omega) = \frac{\pi}{2} - \operatorname{atan2}\!\left(\sqrt{2}\,\omega,\; 1-\omega^2\right) $$

Tipo de filtro y ancho de banda:

$$ \frac{2\omega^2}{\omega^4 + 1} = \frac{1}{2} \;\Rightarrow\; \omega^4 - 4\omega^2 + 1 = 0 \;\Rightarrow\; \omega_{1,2} = \sqrt{2 \mp \sqrt{3}} \approx 0{,}518\ ;\ 1{,}932\ \text{rad/s} $$

$$ \boxed{BW = \omega_2 - \omega_1 = \frac{\omega_0}{Q} = \sqrt{2}\ \text{rad/s}} \qquad (\text{filtro pasabanda}) $$

Funcion de transferencia:

$$ \frac{1}{Z_1} = sC\alpha + \frac{b}{sL} = \frac{s^2 C\alpha L + b}{sL} \;\Rightarrow\; Z_1 = \frac{sL}{s^2 C\alpha L + b} $$

$$ \frac{1}{Z_2} = \frac{1-b}{sL} + \frac{1}{R} = \frac{(1-b)R + sL}{sLR} \;\Rightarrow\; Z_2 = \frac{sLR}{(1-b)R + sL} $$

$$ \frac{1}{Z_2'} = \frac{(1-b)R + sL}{sLR} + sC(1-\alpha) = \frac{(1-b)R + sL + s^2 LCR\,(1-\alpha)}{sLR} \;\Rightarrow\; Z_2' = \frac{sLR}{(1-b)R + sL + s^2 LCR\,(1-\alpha)} $$

$$ H(s) = \frac{Z_2'}{Z_1 + Z_2'} = \frac{R\,(s^2 C\alpha L + b)}{R\,(s^2 C\alpha L + b) + (1-b)R + sL + s^2 LCR\,(1-\alpha)} $$

$$ H(s) = \frac{RC\alpha L\,s^2 + bR}{R + sL + s^2 LCR} = \frac{\alpha\,s^2 + \dfrac{b}{LC}}{s^2 + \dfrac{1}{CR}\,s + \dfrac{1}{LC}} $$

Calculo de L y C, polos y ceros:

$$ LC = 1, \qquad \frac{1}{CR} = \sqrt{2} \;\Rightarrow\; C = \frac{1}{\sqrt{2}}\ \text{F}, \qquad L = \sqrt{2}\ \text{H} $$

$$ \boxed{H(s) = \frac{\alpha\,s^2 + b}{s^2 + \sqrt{2}\,s + 1}} $$

$$ \alpha\,s^2 + b = 0 \;\Rightarrow\; s^2 = -\frac{b}{\alpha} \;\Rightarrow\; s = \pm j\sqrt{\frac{b}{\alpha}} \qquad (\alpha, b > 0) $$
$$ \text{Polos: } p_{1,2} = -\frac{1}{\sqrt{2}} \pm j\,\frac{1}{\sqrt{2}} \qquad  $$

Modulo y fase:

$$ H(\omega) = \frac{-\alpha\,\omega^2 + b}{-\omega^2 + j\sqrt{2}\,\omega + 1} $$

$$ |H(\omega)| = \frac{|b - \alpha\,\omega^2|}{\sqrt{(1-\omega^2)^2 + 2\omega^2}} = \frac{|b - \alpha\,\omega^2|}{\sqrt{\omega^4 + 1}} $$

$$ |H(0)| = b, \qquad |H(1)| = \frac{|b - \alpha|}{\sqrt{2}}, \qquad \lim_{\omega \to \infty} |H(\omega)| = \alpha $$

$$ \angle H(\omega) = \arg\!\left(b - \alpha\,\omega^2\right) - \operatorname{atan2}\!\left(\sqrt{2}\,\omega,\; 1-\omega^2\right), \qquad \arg\!\left(b - \alpha\,\omega^2\right) = \begin{cases} 0 & \omega < \sqrt{b/\alpha} \\ \pi & \omega > \sqrt{b/\alpha} \end{cases} $$

#### Conclusión:
Es posible generar distintos tipos de filtros analógicos con circuitos electrónicos variando la configuración de los componentes de la siguiente forma:

Capacitores en serie e inductores en paralelo refieren a filtros pasa altos, donde las frecuencias bajas serán anuladas.

Capacitores en paralelo e inductores en serie: filtros pasa bajos, donde las frecuencias bajas serán anuladas.

Capacitores e inductores en una misma derivación: filtros pasa banda, donde pasará la información que se encuentre en la frecuencia de corte y sus arlededores.

En los filtros pasa bajos y pasa altos la frecuencia de corte delimita la frecuencia donde el módulo de la función de transferencia alcanza los -3dB. Valores mayores a este comienzan a ser considerados obtenidos y menores a ser descartados. Por esto la frecuencia de corte en estos filtros denota las frecuencias donde sucederá la obtención de información.
