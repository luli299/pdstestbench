En este trabajo lo que hicimos fue crear un generador de señales; como una funcion de señales senoidales y parametrizamos la amplitud maxima de la senoidal (volts), su valor medio (volts), su frecuencia (Hz), la fase (radianes), la cantidad de muestras digitalizadas por el ADC (#muestras) y una frecuencia de muestreo del ADC.


Comenzamos definiendo nuestra frecuencia (fs) y nuestras muestras. Luego implementamos la funcion.

```python
import numpy as np
import matplotlib.pyplot as plt
#cuando hay redun hay una senoidal

#%%definiciones
fs=1000 #Hz de Nyquitz 
N=1000 #muestras
#%%funciones
def mi_funcion_sen( vmax = 1, dc = 0, ff = 1, ph=0, nn = N, fs = fs):
    
    tt=np.arange(N)/fs #se mide en segundos
    xx = vmax * np.sin( 2 * np.pi * ff * tt + ph ) #valores de la senoidal
    #tt=np.arange(N)/fs #se mide en segundos
    
    return(tt, xx)
```
